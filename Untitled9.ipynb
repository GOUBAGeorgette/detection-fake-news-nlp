#NETTOYAGE DU TEXTE

L'ordinateur ne comprend pas le texte brut tel quel.nous allons le nettoyer en supprimant les éléments inutiles comme URLs,mentions, hashtags, chiffres isolés, et nous allons mettre tout en minuscules. Nous allons supprimer aussi les stopwords (mots vides comme "le", "de", "un" qui n'apportent aucune information).

Chargement des stopwords français

```python
import nltk
nltk.download('stopwords', quiet=True)
import re
from nltk.corpus import stopwords

stop_fr = set(stopwords.words('french'))

def nettoyer_texte(texte):
    """
    Nettoie un texte brut pour la vectorisation NLP.
    Étapes : minuscules → URLs → mentions/hashtags →
             ponctuation → espaces → stopwords.
    """
    texte = str(texte).lower()                        # minuscules
    texte = re.sub(r'http\S+|www\S+', '', texte)      # supprime les URLs
    texte = re.sub(r'[@#]\w+', '', texte)             # supprime mentions et hashtags
    texte = re.sub(r'\d+', '', texte)                 # supprime les chiffres isolés
    texte = re.sub(r'[^\w\s]', ' ', texte)            # supprime la ponctuation
    texte = re.sub(r'\s+', ' ', texte).strip()        # supprime les espaces multiples
    # Suppression des stopwords mot par mot
    mots = texte.split()
    mots = [m for m in mots if m not in stop_fr and len(m) > 2]
    return ' '.join(mots)
```

Application du nettoyage

```python
df['texte_propre'] = df['texte'].apply(nettoyer_texte)
```

```python
# --- Affichage avant / après nettoyage ---
print("=" * 50)
print("RÉSULTAT DU NETTOYAGE DU TEXTE")
print("=" * 50)
for i in range(3):
    print(f"\n[Ligne {i+1}]")
    print(f"  AVANT : {df['texte'].iloc[i][:120]}")
    print(f"  APRÈS : {df['texte'].iloc[i][:120]}")

print(f"\nAperçu du dataset avec la colonne texte_propre :")
print(df[['source', 'texte_propre', 'label']].head(10))
```

#FEATURE ENGINEERING (INDICES NUMÉRIQUES)

En plus des mots, nous allons créer des indices numériques qui capturent
le comportement stylistique des fake news : textes trop courts
ou trop longs, abus de majuscules, ponctuation agressive, mots exagérés, mots négatifs

```python
def compter_majuscules(texte):
    """Calcule la proportion de mots entièrement en majuscules."""
    mots = str(texte).split()
    if len(mots) == 0:
        return 0
    mots_maj = [m for m in mots if m.isupper() and len(m) > 1]
    return len(mots_maj) / len(mots)

# Listes de mots pour l'analyse d'impact émotionnel/rhétorique
mots_exageres = ['waouh', 'incroyable', 'exceptionnel', 'extraordinaire', 'magnifique', 'formidable', 'génial', 'fantastique', 'super', 'absolument', 'énorme', 'incroyablement', 'totalement', 'définitivement', 'jamais']
mots_negatifs = ['horrible', 'terrible', 'scandaleux', 'choquant', 'honteux', 'catastrophe', 'danger', 'attention', 'alerte', 'pas bon', 'faux', 'mensonge', 'tragique', 'effrayant', 'désastre', 'pire', 'grave']

def compter_mots_specifiques(texte, liste_mots):
    """
    Compte le nombre de mots d'une liste donnée dans un texte, insensible à la casse.
    """
    texte_lower = str(texte).lower()
    count = 0
    for mot in liste_mots:
        count += texte_lower.count(mot.lower())
    return count
```

Création des nouvelles variables

```python
df['nb_mots']         = df['texte'].apply(lambda x: len(str(x).split()))
df['nb_caracteres']   = df['texte'].apply(lambda x: len(str(x)))
df['prop_majuscules'] = df['texte'].apply(compter_majuscules)
df['nb_exclamation']  = df['texte'].apply(lambda x: str(x).count('!'))
df['nb_interrogation']= df['texte'].apply(lambda x: str(x).count('?'))
df['nb_mots_exageres'] = df['texte'].apply(lambda x: compter_mots_specifiques(x, mots_exageres))
df['nb_mots_negatifs'] = df['texte'].apply(lambda x: compter_mots_specifiques(x, mots_negatifs))
```

```python
# --- Affichage des nouveaux indices ---
print("=" * 50)
print("INDICES NUMÉRIQUES CRÉÉS (FEATURE ENGINEERING)")
print("=" * 50)
print(df[['texte_propre', 'nb_mots', 'nb_caracteres',
          'prop_majuscules', 'nb_exclamation',
          'nb_interrogation', 'nb_mots_exageres', 'nb_mots_negatifs', 'label']].head(8))

print(f"\nMoyenne des indices par classe :")
print(df.groupby('label')[['nb_mots', 'nb_caracteres',
                            'prop_majuscules', 'nb_exclamation',
                            'nb_interrogation', 'nb_mots_exageres', 'nb_mots_negatifs']].mean().round(3))
```

#VECTORISATION TF-IDF

Le TF-IDF transforme chaque texte en un vecteur de nombres.
Chaque nombre représente l'importance d'un mot dans le post par rapport à tout le dataset. max_features=2000 signifie qu'on garde les 2000 mots les plus informatifs.

Initialisation du vectoriseur

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features = 2000,      # on garde les 2000 mots les plus importants
    ngram_range  = (1, 2),    # on analyse les mots seuls ET les paires de mots
    min_df       = 2,         # un mot doit apparaître au moins 2 fois
    sublinear_tf = True       # lissage logarithmique des fréquences
)
```

```python
#---Application du TF-IDF sur le texte nettoyé
X_tfidf = tfidf.fit_transform(df['texte_propre'])

# --- Récupération des indices numériques
features_num = df[['nb_mots', 'nb_caracteres', 'prop_majuscules',
                    'nb_exclamation', 'nb_interrogation', 'nb_mots_exageres', 'nb_mots_negatifs']].values

# --- Fusion TF-IDF + indices numériques ---
from scipy.sparse import hstack, csr_matrix
X = hstack([X_tfidf, csr_matrix(features_num)])
y = df['label'].values

# --- Affichage du résultat ---
print("=" * 50)
print("RÉSULTAT DE LA VECTORISATION TF-IDF")
print("=" * 50)
print(f"\nDimensions de la matrice X : {X.shape}")
print(f"  → {X.shape[0]} posts")
print(f"  → {X.shape[1]} variables (2000 TF-IDF + {len(features_num[0])} indices numériques)")
print(f"\nTop 50 mots les plus importants selon TF-IDF :")
print(tfidf.get_feature_names_out()[:20])
```

#SÉPARATION TRAIN / TEST

On divise le dataset en deux parties :
- 80% pour entraîner les modèles (jeu d'entraînement)
- 20% pour les tester sur des données jamais vues (jeu de test)

# ENTRAÎNEMENT DES 4 MODÈLES

On entraîne les 4 modèles choisis. Chaque modèle apprend
à reconnaître les fake news à partir du jeu d'entraînement.


Définition des 4 modèles

```python
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

modeles = {
    "Régression Logistique" : LogisticRegression(max_iter=1000, random_state=42),
    "Naïve Bayes"           : MultinomialNB(),
    "Random Forest"         : RandomForestClassifier(n_estimators=100, random_state=42),
    "XGBoost"               : XGBClassifier(n_estimators=100, random_state=42,
                                             eval_metric='logloss', verbosity=0)
}
```

#ÉVALUATION ET COMPARAISON DES MODÈLES

Testons chaque modèle sur le jeu de test (données jamais vues).
Calculons 4 métriques pour chaque modèle :
Accuracy, F1-Score, Précision, Rappel.

```python
from sklearn.metrics import (accuracy_score, f1_score, precision_score,
                             recall_score, roc_auc_score, roc_curve)

print("=" * 50)
print("ÉVALUATION DES 4 MODÈLES SUR LE JEU DE TEST")
print("=" * 50)

# Ensure modeles_entraines is defined by re-training models
# (This code is taken from cell 3DREATm7U6Xq)
modeles_entraines = {}
for nom, modele in modeles.items():
    modele.fit(X_train, y_train)
    modeles_entraines[nom] = modele

resultats = []

for nom, modele in modeles_entraines.items():
    y_pred  = modele.predict(X_test)
    y_proba = modele.predict_proba(X_test)[:, 1]

    acc  = accuracy_score(y_test, y_pred)
    f1   = f1_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred)
    rec  = recall_score(y_test, y_pred)
    auc  = roc_auc_score(y_test, y_proba)

    resultats.append({
        'Modèle'    : nom,
        'Accuracy'  : round(acc, 4),
        'F1-Score'  : round(f1, 4),
        'Précision' : round(prec, 4),
        'Rappel'    : round(rec, 4),
        'AUC-ROC'   : round(auc, 4)
    })

    print(f"\n── {nom} ──")
    print(f"   Accuracy  : {acc:.2%}")
    print(f"   F1-Score  : {f1:.2%}")
    print(f"   Précision : {prec:.2%}")
    print(f"   Rappel    : {rec:.2%}")
    print(f"   AUC-ROC   : {auc:.2%}")
```

Tableau comparatif

```python
df_resultats = pd.DataFrame(resultats).sort_values('F1-Score', ascending=False)
print("\n" + "=" * 50)
print("TABLEAU COMPARATIF DES 4 MODÈLES")
print("=" * 50)
print(df_resultats.to_string(index=False))

meilleur = df_resultats.iloc[0]['Modèle']
print(f"\n🏆 Meilleur modèle selon le F1-Score : {meilleur}")

```

```python
import shap
import matplotlib.pyplot as plt
import numpy as np

print("=" * 50)
print("ANALYSE SHAP — INTERPRÉTABILITÉ DU MEILLEUR MODÈLE")
print("=" * 50)

# On construit la liste complète des noms de variables
noms_tfidf = list(tfidf.get_feature_names_out())
noms_num   = ['nb_mots', 'nb_caracteres', 'prop_majuscules',
              'nb_exclamation', 'nb_interrogation', 'nb_mots_exageres', 'nb_mots_negatifs']
noms_vars  = noms_tfidf + noms_num

# Sample X_test for SHAP explanation
num_shap_samples = 200
if X_test.shape[0] > num_shap_samples:
    shap_test_indices = np.random.choice(X_test.shape[0], num_shap_samples, replace=False)
else:
    shap_test_indices = np.arange(X_test.shape[0])
X_test_sampled_dense = X_test[shap_test_indices].toarray()

# For TreeExplainer, it is recommended to use a small, representative subset of the training data
# (or the mean of the training data) as 'background data'.
num_background_samples = 100
if X_train.shape[0] > num_background_samples:
    background_data_indices = np.random.choice(X_train.shape[0], num_background_samples, replace=False)
else:
    background_data_indices = np.arange(X_train.shape[0])
background_data = X_train[background_data_indices].toarray()

# Use shap.TreeExplainer for tree-based models like RandomForestClassifier
explainer = shap.TreeExplainer(meilleur_modele, data=background_data)

# Calculate SHAP values for the sampled test set
shap_values_raw = explainer.shap_values(X_test_sampled_dense)

# Handle the output format of shap_values_raw
# For binary classification, TreeExplainer.shap_values can return either:
# 1. A list of two 2D arrays: [shap_values_class_0, shap_values_class_1]
# 2. A single 3D array: (num_samples, num_features, num_classes)
# Based on previous debugging output, it was a 3D array.
if isinstance(shap_values_raw, list):
    # If it's a list, we expect two elements for binary classification
    if len(shap_values_raw) == 2:
        shap_values_for_plot = shap_values_raw[1] # Take SHAP values for the positive class (label 1)
    else:
        # Fallback if it's a list but not binary, take the first element
        print("Warning: shap_values_raw is a list, but not of length 2. Taking the first element.")
        shap_values_for_plot = shap_values_raw[0]
elif isinstance(shap_values_raw, np.ndarray) and shap_values_raw.ndim == 3 and shap_values_raw.shape[2] == 2:
    # This path is taken if shap_values_raw is directly a 3D array (samples, features, classes)
    shap_values_for_plot = shap_values_raw[:, :, 1] # Take SHAP values for the positive class (index 1)
else:
    # If it's a 2D array directly (e.g., regressor or single-output model), use as is
    shap_values_for_plot = shap_values_raw
    print("Warning: shap_values_raw is a 2D array or unhandled format. Using directly.")


# --- SHAP plot: top 30 most influential variables ---
shap.summary_plot(shap_values_for_plot, X_test_sampled_dense,
                  feature_names=noms_vars,
                  max_display=30,
                  show=False)
plt.title(f'Variables les plus influentes — {meilleur}')
plt.tight_layout()
plt.savefig('shap_importance.png', dpi=150)
plt.show()
```

# TEST SUR UN NOUVEAU POST (DÉMONSTRATION)

Testons le meilleur modèle sur quelques publication rélles q'il n'a jamais rencontré


```python
# ================================================================
# TEST SUR DES POSTS RÉELS — DÉMONSTRATION
# À exécuter après l'entraînement des modèles dans Colab
# ================================================================

from scipy.sparse import hstack, csr_matrix
import numpy as np

# Retrieve the best model
meilleur_modele = modeles_entraines[meilleur]

def predire_post(texte_brut, modele, vectoriseur):
    """
    Prédit si un post est une fake news ou une vraie information.
    Retourne le label prédit et la probabilité associée.
    """
    # Nettoyage du texte
    texte_net = nettoyer_texte(texte_brut)

    # Vectorisation TF-IDF
    vecteur_tfidf = vectoriseur.transform([texte_net])

    # Calcul des indices numériques
    indices = np.array([[
        len(texte_brut.split()),
        len(texte_brut),
        compter_majuscules(texte_brut),
        texte_brut.count('!'),
        texte_brut.count('?'),
        compter_mots_specifiques(texte_brut, mots_exageres),
        compter_mots_specifiques(texte_brut, mots_negatifs)
    ]])

    # Fusion TF-IDF + indices numériques
    vecteur_final = hstack([vecteur_tfidf, csr_matrix(indices)])

    # Prédiction
    pred  = modele.predict(vecteur_final)[0]
    proba = modele.predict_proba(vecteur_final)[0]

    return pred, proba


# --- Publications réelles burkinabè jamais vues par le modèle ---
posts_test = [
    "Triste nouvelle, après la défaite face à l'Allemagne, la maison des parents "
    "de Simon Adingra a été la cible de certains supporters sise à Abobo !!!",

    """Tabaski 2026 : Les fidèles musulmans sunnites ont célébré la prière de l'Aïd
    El-Kébir à Ouaga 2000. La prière a été dirigée par l'imam Mahamoud Ouédraogo,
    en présence de nombreux fidèles venus accomplir la grande prière de l'Aïd El-Kébir.""",

    "Burkina Faso : La banque de Moscou va accueillir les réserves d'or du Burkina. "
    "Le Ministre de l'Économie annonce la signature d'un Protocole d'Accord entre "
    "le Burkina Faso et la Banque de Moscou relatif au stockage des réserves d'or !!!",
]

# --- Affichage des résultats ---
print("=" * 60)
print("TEST DE DÉMONSTRATION SUR DES PUBLICATIONS RÉELLES")
print("=" * 60)

for i, post in enumerate(posts_test, 1):
    pred, proba = predire_post(post, meilleur_modele, tfidf)

    verdict   = "🔴 FAKE NEWS"       if pred == 1 else "🟢 INFORMATION VRAIE"
    confiance = proba[pred] * 100
    p_fake    = proba[1] * 100
    p_vrai    = proba[0] * 100

    print(f"\n[Post {i}]")
    print(f"  Texte     : {post[:90]}...")
    print(f"  Verdict   : {verdict}")
    print(f"  Confiance : {confiance:.1f}%")
    print(f"  P(Vrai)   : {p_vrai:.1f}%  |  P(Fake) : {p_fake:.1f}%")
    print(f"  Nb mots   : {len(post.split())}  |  Exclamations : {post.count('!')}")
```
